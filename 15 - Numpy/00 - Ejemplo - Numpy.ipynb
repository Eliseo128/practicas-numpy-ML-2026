### Diferencia entre listas y arreglos

````python
# Diferencia entre listas y arreglos de numpy

# Lista
lista = [1, "A", [1,2,3]]
print(lista)
print(type(lista))
````

````python
# Arreglo
arr = np.array([1, 2, 3])
print(arr)
print(type(arr))
````

````python
# Revisar métodos de las listas
lista_metodos = [i for i in dir(list()) if "_" not in i]
print(lista_metodos)
````

````python
# Revisar métodos de arreglos
arreglos_metodos = [i for i in dir(np) if "_" not in i and i.islower()]
print(arreglos_metodos)
````

````python
# Tipo del arreglo
arr = np.array([4, 3, 2])
print(arr.dtype) # int32
````

````python
arr = np.array([4, 3.0, 2])
print(arr.dtype) # float64
````

````python
arr = np.array([4, '3', 2])
print(arr.dtype) # <U11 tipo de cadena de texto
````

````python
# Operaciones
lista = [1,2,3]
arr = np.array([1,2,3])

print(lista*3)
print(arr*3)
````

## Crear arreglos

````python
# Crear arreglo vacío
np.empty(3)
````

````python
# Crear arreglo de 0's
np.zeros((3,2))
````

````python
# Crear arreglo de 1's
np.ones((3,2))
````

````python
# Crear arreglo con un valor definido
np.full((3,2),7)
````

````python
# Crear matriz identidad
np.eye(3)
````

````python
# Crear arreglo
np.linspace(start=1, stop=10, num=5) #, endpoint=False)
````

````python
# Crear arreglo
np.arange(start=0, stop=8, step=2)
````

### Crear arreglos de números aleatorios

````python
# Número aleatorio en distribución uniforme
np.random.rand(2,3)
````

````python
# Número aleatorio entre un rango
np.random.randint(20, size=(2,4))
````

````python
# Número aleatorio en distribución normal
np.random.normal(size=(2,5))
````

````python
# Revisar media en distribución normal (muy cercana a 0)
np.random.normal(size=(1000,1000)).mean()
````

````python
# Revisar varianza en distribución normal (muy cercana a 1)
np.random.normal(size=(100,100)).var()
````

````python
# Utilizar semillas para obtener siempre el mismo valor aleatorio
np.random.seed(0)
np.random.rand(2,3)
````

### Filtrar arreglos

````python
# Filtros
arr = np.array([1,2,3,4,5,6])
print(arr.shape)
print()
print(arr)
````

````python
filtro = np.array([False, True, False, True, False, False])
print(arr)
print(filtro)
print(arr[filtro])
````

````python
filtro = (arr > 3) & (arr < 6)
print(arr)
print(filtro)
print(arr[filtro])
````

````python
print(arr[arr > 3])
````
